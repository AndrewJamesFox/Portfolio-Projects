# Understanding the data

The dataset is from https://uni-siegen.sciebo.de/s/HGdUkoNlW1Ub0Gx from https://archive.ics.uci.edu/dataset/465/wesad+wearable+stress+and+affect+detection

According to the documentation, the data is stored as a dictionary with the following keys:
- 'subject' : the subject ID
- 'signal' : includes the raw signal data, in two fields according to device:
    - 'chest' : chest device data, which includes these signals : ACC, ECG, EDA, EMG, RESP, TEMP, all sampled at 700hz
    - 'wrist' : wrist device data, which includes these signals : ACC, BVP, EDA, TEMP, sampled at 32hz, 64hz, 4hz, 4hz, respectively
- 'label' : the label ID for the respective protocol condition, which are as follows: 0 = not defined | 1 = baseline | 2 = stress | 3 = amusement | 4 = meditation | 5,6,7 should be ignored. The labels are aligned to the 700hz chest sampling rate.

The raw data is stored in my raw data dictionary, whose path is stored in RAWDIR. In RAWDIR, there are 15 subject directories named 'SX', where X is the subject number. In each there is a .pkl dictionary named 'sX' with the synchronized data from both the chest and wrist devices. I will load each subject's data into a .pkl which I will save in RAWDIR for access across this project.

# Getting the data

```python
#initialize rawdata dict
all_rawdata = {}

# skip files that aren't subject data directories
sub_dirs = [d for d in os.listdir(RAWDIR) if d.startswith("S")]

for sub_dir in sorted(sub_dirs, key=lambda x: int(x[1:])):
    subnum = int(sub_dir[1:]) #retrieve subject number
    pkl_path = os.path.join(RAWDIR, sub_dir, f"{sub_dir}.pkl") #get path to subject's .pkl file

    # load subject's .pkl into all_raw_data dict
    if os.path.exists(pkl_path):
        with open(pkl_path, "rb") as f:
            all_rawdata[subnum] = pickle.load(f, encoding="latin1")
```

```python
# Save all_raw_data to RAWDIR for future use
with open(RAWDATA, "wb") as f:
    pickle.dump(all_rawdata, f)
```

The raw subject data can now be retrieved across the project with this function from src.helpers:
